Before approximating an ordinary differential equation numerically, we should ask
whether the initial-value problem has a solution, whether that solution is unique,
and whether small perturbations lead only to small changes in the solution.

Consider the first-order IVP

$$
y'=f(t,y), \qquad a\le t\le b, \qquad y(a)=\alpha.
$$

### Learning goals

- identify an initial-value problem (IVP);
- test continuity of $f(t,y)$;
- check a Lipschitz condition in $y$;
- estimate a Lipschitz constant;
- apply the existence-and-uniqueness theorem used in the course;
- explain what it means for an IVP to be well-posed;
- recognize how perturbations connect theory to numerical computation.


## 1. Lipschitz continuity in $y$

A function $f(t,y)$ satisfies a **Lipschitz condition in $y$** on a set
$D\subset\mathbb{R}^2$ if there is a constant $L>0$ such that

$$
|f(t,y_1)-f(t,y_2)| \le L|y_1-y_2|
$$

whenever $(t,y_1)$ and $(t,y_2)$ lie in $D$.

A practical test used in the course is:

> If $D$ is convex and $\partial f/\partial y$ is continuous and bounded on
> $D$, then $f$ is Lipschitz in $y$ on $D$.

A valid Lipschitz constant is any number $L$ satisfying

$$
\left|\frac{\partial f}{\partial y}(t,y)\right|\le L
$$

throughout the domain.


```python
import numpy as np


def max_abs_fy(f_y, t_values, y_values):
    """Estimate a bound for |partial f / partial y| on a grid."""
    t_grid, y_grid = np.meshgrid(t_values, y_values)
    return np.max(np.abs(f_y(t_grid, y_grid)))
```

## 2. Review-style examples

The Exam 3 review asks whether several functions satisfy a Lipschitz condition on

$$
D=\{(t,y):0\le t\le1,\ -\infty<y<\infty\}.
$$

### Example A

$$
f(t,y)=t^2y+1,
\qquad
\frac{\partial f}{\partial y}=t^2.
$$

Since $0\le t\le1$,

$$
\left|\frac{\partial f}{\partial y}\right|=t^2\le1.
$$

Therefore $L=1$ works.

### Example B

$$
f(t,y)=1-y,
\qquad
\frac{\partial f}{\partial y}=-1.
$$

Again, $L=1$ works.

### Example C

$$
f(t,y)=ty,
\qquad
\frac{\partial f}{\partial y}=t.
$$

On $0\le t\le1$, $L=1$ works.

The domain can be unbounded in $y$ and still have a global Lipschitz constant if
$\partial f/\partial y$ remains bounded.


```python
t_values = np.linspace(0.0, 1.0, 101)
y_values = np.linspace(-10.0, 10.0, 201)

examples = {
    "t^2 y + 1": lambda t, y: t**2,
    "1 - y": lambda t, y: -np.ones_like(t),
    "t y": lambda t, y: t,
}

for name, f_y in examples.items():
    estimate = max_abs_fy(f_y, t_values, y_values)
    print(f"{name:12s}: estimated L = {estimate:.6f}")
```

## 3. Existence and uniqueness

A useful version of the theorem used in the course is:

> If $f(t,y)$ is continuous on a suitable domain containing the initial point
> and satisfies a Lipschitz condition in $y$, then the IVP has a unique solution
> on the interval covered by the theorem.

The two checks play different roles:

- continuity of $f$ supports **existence**;
- the Lipschitz condition in $y$ supports **uniqueness**.

Continuity by itself does **not** guarantee uniqueness.


## 4. Where the conditions fail

### No solution at the initial point

$$
y'=\frac{1}{t}, \qquad y(0)=0.
$$

The function $f(t,y)=1/t$ is not defined at $t=0$, so the continuity hypothesis
fails at the initial point.

### Nonunique behavior

$$
y'=3y^{2/3}, \qquad y(0)=0.
$$

Formally,

$$
\frac{\partial f}{\partial y}=2y^{-1/3},
$$

which is unbounded near $y=0$. The practical Lipschitz test fails there.

Failure of the theorem does not automatically prove nonexistence or
nonuniqueness. It means the theorem can no longer guarantee the conclusion.


## 5. Well-posedness and perturbations

An IVP is **well-posed** when:

1. a solution exists;
2. the solution is unique;
3. the solution depends continuously on the data.

The third condition matters numerically. A computer works with approximations,
so small changes in the initial value or in the differential equation should not
produce arbitrarily large changes in the solution.

A perturbed problem has the form

$$
z'=f(t,z)+\delta(t),
\qquad
z(a)=\alpha+\varepsilon_0.
$$

Well-posedness asks whether small $\delta(t)$ and small $\varepsilon_0$ produce a
solution $z(t)$ that remains close to the original solution $y(t)$.


## 6. Exam-problem checklist

For an IVP

$$
y'=f(t,y), \qquad y(t_0)=y_0,
$$

use this sequence:

1. **State the domain** $D$.
2. **Check continuity** of $f$ on $D$.
3. Compute $\partial f/\partial y$.
4. Find a bound $|\partial f/\partial y|\le L$ on $D$.
5. Conclude that $f$ is Lipschitz in $y$.
6. Apply the existence-and-uniqueness theorem.
7. If asked about well-posedness, explain continuous dependence on perturbations.
