# Section 4 — Fixed-Point Iteration and Cobweb Diagrams

A fixed point of $g$ is a number $p$ satisfying

$$
g(p)=p.
$$

Fixed-point iteration starts with $p_0$ and defines

$$
p_n=g(p_{n-1}),\qquad n\ge1.
$$

A cobweb diagram shows this iteration geometrically by alternating between the graph
$y=g(x)$ and the diagonal $y=x$.

### Learning goals
- identify fixed points;
- carry out fixed-point iteration;
- draw and interpret a cobweb diagram;
- connect the picture to the condition $|g'(x)|<1$;
- use the Fixed-Point Theorem to guarantee convergence.

```python
import math

import matplotlib.pyplot as plt
import numpy as np
```

## Example used in the course

Consider

$$
g(x)=\frac{5}{x+4}.
$$

The fixed points satisfy

$$
x=\frac{5}{x+4},
$$

so $x=1$ and $x=-5$. Starting from $p_0=1.5$, the iteration converges to
the fixed point $p=1$.

```python
def g(x):
    return 5.0 / (x + 4.0)


p = 1.0
p_n = 1.5
iterates = [p_n]

for _ in range(8):
    p_n = g(p_n)
    iterates.append(p_n)

print(" n          p_n           |p_n - p|")
for n, value in enumerate(iterates):
    error = abs(value - p)
    print(f"{n:2d}   {value:13.10f}   {error:13.6e}")
```

## Cobweb diagram

The construction is:

1. start at $(p_0,p_0)$ on $y=x$;
2. move vertically to $y=g(x)$;
3. move horizontally back to $y=x$;
4. repeat.

When the steps spiral inward toward the intersection, the iteration converges.

```python
def cobweb(function, p0, x_min, x_max, iterations=8):
    """Plot a cobweb diagram for p_n = g(p_{n-1})."""
    x_values = np.linspace(x_min, x_max, 500)

    plt.figure(figsize=(7, 7))
    plt.plot(x_values, x_values, label=r"$y=x$")
    plt.plot(x_values, function(x_values), label=r"$y=g(x)$")

    p_old = p0
    for _ in range(iterations):
        p_new = function(p_old)
        plt.plot([p_old, p_old], [p_old, p_new], linewidth=1.5)
        plt.plot([p_old, p_new], [p_new, p_new], linewidth=1.5)
        p_old = p_new

    plt.scatter([p], [p], zorder=3, label="fixed point p = 1")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Cobweb Diagram for g(x) = 5 / (x + 4)")
    plt.grid(alpha=0.25)
    plt.legend()
    plt.axis("equal")
    plt.show()


cobweb(g, 1.5, 0.5, 1.6)
```

## Why does it converge?

On an interval [a,b], the Fixed-Point Theorem used in the course asks us to check:

1. $g$ is continuous on [a,b];
2. $g(x)\in[a,b]$ for every $x\in[a,b]$;
3. there is a constant $0<k<1$ such that $|g'(x)|\le k$ on $(a,b)$.

Then the fixed point is unique and the iteration converges for every starting value in
the interval.

For this example,

$$
g'(x)=-\frac{5}{(x+4)^2}.
$$

```python
a = 0.5
b = 1.5

g_a = g(a)
g_b = g(b)
k = 5.0 / (a + 4.0) ** 2

image_min = min(g_a, g_b)
image_max = max(g_a, g_b)
print(f"g([{a}, {b}]) lies between {image_min:.6f} and {image_max:.6f}")
print(f"maximum derivative magnitude bound k = {k:.6f}")
print(f"k < 1: {k < 1}")
```

## Iteration-count estimate

The contraction estimate

$$
|p_n-p|\le k^n |p_0-p|
$$

gives a simple way to estimate how many iterations are sufficient for a requested
accuracy.

```python
tolerance = 1.0e-5
p0 = 1.5
initial_error = abs(p0 - p)

n_required = math.ceil(
    math.log(tolerance / initial_error) / math.log(k)
)

print(f"sufficient number of iterations = {n_required}")
```
