Sullivan's numerical-calculus chapter begins integration with the Riemann-sum
idea:

$$
\int_a^b f(x)\,dx
\approx
\sum_{j=1}^{N} f(x_j)\Delta x.
$$

Riemann sums approximate the function on each subinterval by a **constant**.

Common choices are:

- left endpoint;
- right endpoint;
- midpoint.

For a smooth function, left and right sums are typically first-order methods,
while the midpoint rule is second order.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def f(x):
    return x**2


def left_riemann(function, a, b, n):
    h = (b - a) / n
    x = a + h * np.arange(n)
    return h * np.sum(function(x))


def right_riemann(function, a, b, n):
    h = (b - a) / n
    x = a + h * np.arange(1, n + 1)
    return h * np.sum(function(x))


def midpoint_riemann(function, a, b, n):
    h = (b - a) / n
    x = a + h * (np.arange(n) + 0.5)
    return h * np.sum(function(x))
```

## Example: integrate $f(x)=x^2$ on $[0,2]$

The exact value is

$$
\int_0^2 x^2\,dx=\frac83.
$$

```python
a = 0.0
b = 2.0
n = 8
exact = 8.0 / 3.0

left = left_riemann(f, a, b, n)
right = right_riemann(f, a, b, n)
midpoint = midpoint_riemann(f, a, b, n)

print(f"exact     = {exact:.10f}")
print(f"left      = {left:.10f}")
print(f"right     = {right:.10f}")
print(f"midpoint  = {midpoint:.10f}")
```

## Visualize the rectangles

```python
h = (b - a) / n
left_edges = a + h * np.arange(n)
midpoints = left_edges + h / 2.0

x_plot = np.linspace(a, b, 500)

plt.figure(figsize=(10, 5))
plt.plot(x_plot, f(x_plot), linewidth=2, label=r"$f(x)=x^2$")
plt.bar(
    left_edges,
    f(midpoints),
    width=h,
    align="edge",
    alpha=0.35,
    edgecolor="black",
    label="midpoint rectangles",
)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Midpoint Riemann Sum")
plt.grid(alpha=0.2)
plt.legend()
plt.show()
```

## Compare the order

We compute the error as the number of subintervals doubles.

```python
n_values = 2 ** np.arange(2, 10)

left_errors = []
midpoint_errors = []

for count in n_values:
    left_errors.append(
        abs(exact - left_riemann(f, a, b, count))
    )
    midpoint_errors.append(
        abs(exact - midpoint_riemann(f, a, b, count))
    )

plt.figure(figsize=(9, 5))
plt.loglog(n_values, left_errors, "o-", label="left sum")
plt.loglog(n_values, midpoint_errors, "o-", label="midpoint sum")
plt.xlabel("number of subintervals n")
plt.ylabel("absolute error")
plt.title("Riemann-Sum Error")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

### Interpretation

Increasing $n$ decreases $\Delta x$. A first-order method has error proportional
to $\Delta x$, while a second-order method has error proportional to
$(\Delta x)^2$.
