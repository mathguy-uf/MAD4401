# Section 1 — Lagrange Interpolation and Error

[← Module 2](README.md) · [Next: LIP vs. Taylor →](02_lip_vs_taylor.ipynb)

Interpolation asks for a function that passes **exactly** through prescribed data.
For $n+1$ distinct nodes $x_0,\ldots,x_n$, there is a unique polynomial of degree
at most $n$ that interpolates the data.

The Lagrange form is

$$
P_n(x)=\sum_{i=0}^n f(x_i)L_i(x),
$$

where

$$
L_i(x)=\prod_{\substack{j=0\\j\ne i}}^n
\frac{x-x_j}{x_i-x_j}.
$$

Each basis polynomial satisfies $L_i(x_i)=1$ and $L_i(x_j)=0$ for $j\ne i$.

### Learning goals
- construct a Lagrange interpolating polynomial (LIP);
- evaluate it without expanding the polynomial;
- understand why the interpolant has degree **at most** $n$;
- use the interpolation remainder to bound the error.

```python
import math

import matplotlib.pyplot as plt
import numpy as np
```

## Course-style example

Let

$$
f(x)=\sqrt{x},
$$

with nodes

$$
x_0=\frac{1}{9},\qquad
x_1=\frac{1}{4},\qquad
x_2=1.
$$

We will construct the degree-at-most-two interpolating polynomial and use it to
approximate $f(0.5)$.

```python
def lagrange_value(x_nodes, y_nodes, x_value):
    """Evaluate the Lagrange interpolating polynomial at x_value."""
    total = 0.0

    for i, x_i in enumerate(x_nodes):
        basis = 1.0

        for j, x_j in enumerate(x_nodes):
            if i != j:
                basis *= (x_value - x_j) / (x_i - x_j)

        total += y_nodes[i] * basis

    return total


x_nodes = np.array([1.0 / 9.0, 1.0 / 4.0, 1.0])
y_nodes = np.sqrt(x_nodes)
x_value = 0.5

approximation = lagrange_value(x_nodes, y_nodes, x_value)
exact_value = math.sqrt(x_value)
actual_error = abs(exact_value - approximation)

print(f"P2(0.5)      = {approximation:.10f}")
print(f"sqrt(0.5)    = {exact_value:.10f}")
print(f"actual error = {actual_error:.10f}")
```

## Interpolation remainder

If $f\in C^{n+1}$, then for each $x$ there is a point $\xi$ in the interval
containing the nodes and $x$ such that

$$
f(x)-P_n(x)=
\frac{f^{(n+1)}(\xi)}{(n+1)!}
\prod_{i=0}^{n}(x-x_i).
$$

For the quadratic example,

$$
f(x)-P_2(x)=
\frac{f^{(3)}(\xi)}{3!}
(x-x_0)(x-x_1)(x-x_2).
$$

For $f(x)=\sqrt{x}$,

$$
f^{(3)}(x)=\frac{3}{8x^{5/2}}.
$$

On $[1/9,1]$, this derivative decreases, so its maximum magnitude occurs at
$x=1/9$.

```python
x_min = 1.0 / 9.0
m_value = 3.0 / (8.0 * x_min ** 2.5)

product = np.prod(x_value - x_nodes)
error_bound = m_value * abs(product) / math.factorial(3)

print(f"M             = {m_value:.10f}")
print(f"error bound   = {error_bound:.10f}")
print(f"actual error  = {actual_error:.10f}")
```

```python
x_plot = np.linspace(0.1, 1.05, 500)
p_plot = np.array(
    [lagrange_value(x_nodes, y_nodes, x) for x in x_plot]
)

plt.figure(figsize=(9, 5))
plt.plot(x_plot, np.sqrt(x_plot), label=r"$f(x)=\sqrt{x}$")
plt.plot(x_plot, p_plot, label=r"$P_2(x)$")
plt.scatter(x_nodes, y_nodes, zorder=3, label="interpolation nodes")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Lagrange Interpolation")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```
