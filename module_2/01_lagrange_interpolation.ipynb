For $n+1$ distinct nodes $x_0,\ldots,x_n$, there is a **unique**
polynomial of degree at most $n$ that interpolates the data.

The Lagrange form is

$$
P_n(x)=\sum_{i=0}^n f(x_i)L_i(x),
$$

where

$$
L_i(x)=
\prod_{\substack{j=0\\j\ne i}}^n
\frac{x-x_j}{x_i-x_j}.
$$

### Learning goals

- construct a Lagrange interpolating polynomial (LIP);
- explain why the interpolating polynomial exists and is unique;
- understand why its degree is **at most** $n$, not necessarily exactly $n$;
- evaluate an interpolant without expanding it;
- use the interpolation remainder to bound error.


```python
import math

import matplotlib.pyplot as plt
import numpy as np
```

## 1. Why the interpolating polynomial is unique

The Lagrange formula explicitly constructs a degree-at-most-$n$ polynomial,
so existence is established.

For uniqueness, suppose $P_n$ and $Q_n$ are two polynomials of degree at most
$n$ that interpolate the same $n+1$ data values.

Then

$$
R(x)=P_n(x)-Q_n(x)
$$

has degree at most $n$, but

$$
R(x_0)=R(x_1)=\cdots=R(x_n)=0.
$$

Thus $R$ has $n+1$ distinct roots. A nonzero polynomial of degree at most $n$
cannot have more than $n$ roots. Therefore

$$
R(x)\equiv0,
$$

so

$$
\boxed{P_n(x)=Q_n(x)}.
$$


## 2. Course-style example

Let

$$
f(x)=\sqrt{x},
$$

with nodes

$$
x_0=\frac19,\qquad
x_1=\frac14,\qquad
x_2=1.
$$

Construct the degree-at-most-two interpolating polynomial and use it to
approximate $f(0.5)$.


```python
def lagrange_value(x_nodes, y_nodes, x_value):
    """Evaluate the Lagrange interpolating polynomial."""
    total = 0.0

    for i, x_i in enumerate(x_nodes):
        basis = 1.0

        for j, x_j in enumerate(x_nodes):
            if i != j:
                basis *= (x_value - x_j) / (x_i - x_j)

        total += y_nodes[i] * basis

    return total


x_nodes = np.array([1.0 / 9.0, 1.0 / 4.0, 1.0])
y_nodes = np.sqrt(x_nodes)
x_value = 0.5

approximation = lagrange_value(
    x_nodes,
    y_nodes,
    x_value,
)
exact_value = math.sqrt(x_value)
actual_error = abs(exact_value - approximation)

print(f"P2(0.5)      = {approximation:.10f}")
print(f"sqrt(0.5)    = {exact_value:.10f}")
print(f"actual error = {actual_error:.10f}")
```

## 3. Interpolation remainder

If $f\in C^{n+1}$, then for each $x$ there is a point $\xi$ in the interval
containing the nodes and $x$ such that

$$
f(x)-P_n(x)
=
\frac{f^{(n+1)}(\xi)}{(n+1)!}
\prod_{i=0}^{n}(x-x_i).
$$

For $f(x)=\sqrt{x}$,

$$
f^{(3)}(x)=\frac{3}{8x^{5/2}}.
$$

On $[1/9,1]$, this derivative decreases, so its maximum magnitude occurs at
$x=1/9$.


```python
x_min = 1.0 / 9.0
m_value = 3.0 / (8.0 * x_min**2.5)

product = np.prod(x_value - x_nodes)
error_bound = (
    m_value
    * abs(product)
    / math.factorial(3)
)

print(f"M            = {m_value:.10f}")
print(f"error bound  = {error_bound:.10f}")
print(f"actual error = {actual_error:.10f}")
```

## 4. Degree at most $n$ does not mean degree exactly $n$

Four points allow an interpolating polynomial of degree at most three.
However, consider

$$
(1,1),\ (2,3),\ (3,5),\ (4,7).
$$

All four points lie on

$$
P(x)=2x-1.
$$

So the unique interpolating polynomial has degree one, even though four nodes
were supplied.

This is why the theorem says **degree at most $n$**.


```python
x_line_nodes = np.array([1.0, 2.0, 3.0, 4.0])
y_line_nodes = np.array([1.0, 3.0, 5.0, 7.0])

x_plot = np.linspace(0.5, 4.5, 300)
y_plot = np.array(
    [
        lagrange_value(
            x_line_nodes,
            y_line_nodes,
            x_value,
        )
        for x_value in x_plot
    ]
)

plt.figure(figsize=(8, 4.5))
plt.plot(x_plot, y_plot, label="interpolating polynomial")
plt.scatter(
    x_line_nodes,
    y_line_nodes,
    zorder=3,
    label="data",
)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Four Nodes, but Only a Degree-One Interpolant")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```
